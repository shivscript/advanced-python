
## 1. CSV Files (Structured Data)


```python
data = [
    ["id", "name", "age", "city"],
    [1, "Alice", 25, "New York"],
    [2, "Bob", 30, "London"],
    [3, "Charlie", 28, "Sydney"],
    [4, "David", 35, "Toronto"],
    [5, "Eva", 22, "Berlin"],
    [6, "Frank", 40, "Paris"],
    [7, "Grace", 27, "Rome"],
    [8, "Harry", 33, "Tokyo"],
    [9, "Ivy", 29, "Singapore"],
    [10, "Jack", 45, "Dubai"]
]

with open("large_people.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("large_people.csv created")
```

```python

with open("large_people.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```


## 2. JSON Files (Semi-Structured Data)


```python
person = {
    "name": "Alice",
    "age": 25,
    "skills": ["Python", "Data Science"]
}

with open("person.json", "w") as f:
    json.dump(person, f, indent=4)

print("person.json created")
```

```python
with open("person.json", "r") as f:
    data = json.load(f)

print(data)
```


## 3. Excel Files


```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Score": [85, 90, 88]
})

df.to_excel("scores.xlsx", index=False)
print("scores.xlsx created")
```

```python
df_read = pd.read_excel("scores.xlsx")
print(df_read)
```


## 4. Text Files (Unstructured Data)


```python
with open("sample.txt", "w") as f:
    f.write("Python is powerful\n")
    f.write("Advanced Python Programming\n")
    f.write("Handling data sources\n")

print("sample.txt created")
```

```python
with open("sample.txt", "r") as f:
    for line in f:
        print(line.strip())
```


## 5. Chunking Large CSV Files

Chunking allows processing large datasets **without loading the entire file into memory**.


```python
# Reading CSV in chunks
for chunk in pd.read_csv("large_people.csv", chunksize=3):
    print("New Chunk")
    print(chunk)
    print("-" * 30)
```

```python
# Example: counting total rows using chunking
total_rows = 0
for chunk in pd.read_csv("large_people.csv", chunksize=4):
    total_rows += len(chunk)

print("Total rows:", total_rows)
```
