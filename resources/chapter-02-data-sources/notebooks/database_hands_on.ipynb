
## 1. Relational Database Access (SQLite)

SQLite is a lightweight relational database and is ideal for learning SQL concepts.


```python
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
print("Connected to SQLite database")
```

```python
cursor.execute(
    "CREATE TABLE IF NOT EXISTS students ("
    "id INTEGER PRIMARY KEY, "
    "name TEXT, "
    "age INTEGER, "
    "major TEXT)"
)
conn.commit()
print("Table created")
```

```python
students = [
    (1, "Alice", 21, "Computer Engineering"),
    (2, "Bob", 22, "Electronics"),
    (3, "Charlie", 20, "Computer Engineering"),
    (4, "David", 23, "Civil Engineering")
]

cursor.executemany(
    "INSERT OR IGNORE INTO students VALUES (?, ?, ?, ?)",
    students
)
conn.commit()
print("Records inserted")
```

```python
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT name, major FROM students WHERE age > 21")
print(cursor.fetchall())
```

```python
cursor.execute("UPDATE students SET age = age + 1 WHERE name = 'Alice'")
conn.commit()

cursor.execute("SELECT * FROM students WHERE name = 'Alice'")
print(cursor.fetchone())
```

```python
cursor.execute("DELETE FROM students WHERE name = 'David'")
conn.commit()

cursor.execute("SELECT * FROM students")
print(cursor.fetchall())
```

```python
conn.close()
print("Database connection closed")
```


## 2. Non-Relational Database (NoSQL) — Conceptual Hands-on

Here we simulate **document-based NoSQL behavior** similar to MongoDB.


```python
students_collection = [
    {"_id": 1, "name": "Alice", "age": 21, "skills": ["Python", "SQL"]},
    {"_id": 2, "name": "Bob", "age": 22, "skills": ["C++", "Electronics"]},
    {"_id": 3, "name": "Charlie", "age": 20, "skills": ["Python", "ML"]}
]

print(students_collection)
```

```python
python_students = [
    student for student in students_collection
    if "Python" in student["skills"]
]

print(python_students)
```

```python
students_collection.append({
    "_id": 4,
    "name": "Eva",
    "age": 23,
    "skills": ["Data Science", "Python"]
})

print(students_collection)
```

```python
for student in students_collection:
    if student["name"] == "Alice":
        student["age"] = 22

print(students_collection)
```


## 3. OPTIONAL: Connecting Python with MongoDB

```python

# OPTIONAL: Install pymongo if not installed
# !pip install pymongo
```

```python

# OPTIONAL: MongoDB connection example (DO NOT RUN if MongoDB is not installed)

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["college_db"]
collection = db["students"]

print("Connected to MongoDB")
```

```python

# OPTIONAL: Insert a document
collection.insert_one({
    "name": "Alice",
    "age": 21,
    "skills": ["Python", "ML"]
})
```

```python

# OPTIONAL: Query documents
for doc in collection.find({"age": {"$gt": 20}}):
    print(doc)
```

```python

# OPTIONAL: Update a document
collection.update_one(
    {"name": "Alice"},
    {"$set": {"age": 22}}
)
```

```python

# OPTIONAL: Delete a document
collection.delete_one({"name": "Alice"})
```
